# Online Retail Customer Segmentation

## STAGE 1 — Frame & KPIs

**Business question:** How can we segment online retail customers based on purchasing behavior (Recency, Frequency and Monetary value) to identify **High-Value / Champions**, **At-Risk Customers**, and **Low-Engagement Customers** for targeted marketing?

**Decision-maker:** Head of Marketing / CRM Manager. This stakeholder will use the customer segments to decide which groups should receive loyalty benefits, retention campaigns, reactivation offers or low-cost automated communication.

**Business decision supported by this project:** Allocate marketing budget by segment instead of sending the same campaign to every customer.

**KPIs and success metrics:**

- **Revenue by segment:** which segments generate the largest monetary value.
- **Number of customers by segment:** how large each marketing audience is.
- **Average Recency:** how recently each segment purchased.
- **Average Frequency:** how often each segment purchases.
- **Average Monetary value:** how much each segment spends.
- **Technical quality check:** compare different values of `k` using inertia and silhouette score, then compare K-Means with a simple rule-based RFM baseline.

**Modeling approach:** This is an **unsupervised learning** problem because the dataset has no target column such as `Churn`, `Default`, or `Fraud`. We use K-Means clustering to discover customer groups from RFM behavior.


```python
# Import the main libraries used for data analysis, clustering and visualization.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
plt.rcParams["figure.figsize"] = (8, 4)
plt.rcParams["axes.grid"] = True
```

## Load the dataset

This notebook expects `Online Retail.xlsx` in the same folder. The path below is relative, so the notebook runs after `Restart & Run All` with no absolute local paths.

```python
# Load the dataset and set a fixed seed so the results are reproducible.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set random seed for reproducibility 
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

# Load the dataset using a relative path
df = pd.read_csv('Online_Retail.csv')

# Check that the data loaded correctly
df.head()
```

# STAGE 2 — Prepare Data

Each row in the raw dataset represents a product line inside an invoice. For customer segmentation, we need to transform transaction-level data into customer-level features.

Main data preparation choices:

- Remove exact duplicate rows.
- Remove rows without `CustomerID`, because they cannot be assigned to a customer segment.
- Remove returns/cancellations and invalid prices by keeping only `Quantity > 0` and `UnitPrice > 0`.
- Create `TotalAmount = Quantity × UnitPrice`.
- Aggregate by `CustomerID` to create RFM features: `Recency`, `Frequency`, and `Monetary`.
- Avoid leakage by using only historical purchasing behavior, not future campaign results or manually assigned segment labels.

```python
# First look at the raw dataset: columns, missing values and basic statistics.

print("Columns:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isna().sum())

print("\nBasic numeric description:")
df.describe().T
```

```python
# Clean the transaction data before creating customer-level features.

# Cleaning
original_rows = len(df)

df_clean = df.drop_duplicates().copy()
after_duplicates = len(df_clean)

df_clean = df_clean.dropna(subset=["CustomerID"]).copy()
after_customer_filter = len(df_clean)

df_clean = df_clean[(df_clean["Quantity"] > 0) & (df_clean["UnitPrice"] > 0)].copy()
after_returns_filter = len(df_clean)

df_clean["InvoiceDate"] = pd.to_datetime(df_clean["InvoiceDate"])
df_clean["CustomerID"] = df_clean["CustomerID"].astype(int).astype(str)
df_clean["TotalAmount"] = df_clean["Quantity"] * df_clean["UnitPrice"]

cleaning_summary = pd.DataFrame({
    "Step": [
        "Original rows",
        "After removing exact duplicates",
        "After removing missing CustomerID",
        "After removing returns/invalid prices",
    ],
    "Rows": [original_rows, after_duplicates, after_customer_filter, after_returns_filter],
})
cleaning_summary["Rows removed from previous step"] = cleaning_summary["Rows"].shift(1) - cleaning_summary["Rows"]
cleaning_summary
```

```python
# Check the cleaned dataset to confirm that the main filters worked.

print("Clean dataset shape:", df_clean.shape)
print("Date range:", df_clean["InvoiceDate"].min(), "to", df_clean["InvoiceDate"].max())
print("Unique customers:", df_clean["CustomerID"].nunique())
print("Unique invoices:", df_clean["InvoiceNo"].nunique())
print("Unique products:", df_clean["StockCode"].nunique())
print("Total revenue:", round(df_clean["TotalAmount"].sum(), 2))
```

```python
# Create simple EDA charts to understand the business context before modeling.

# Simple EDA charts for the business context
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

top_countries_tx = df_clean["Country"].value_counts().head(10).sort_values()
top_countries_tx.plot(kind="barh", ax=axes[0])
axes[0].set_title("Top 10 countries by transactions")
axes[0].set_xlabel("Transaction lines")

country_revenue = df_clean.groupby("Country")["TotalAmount"].sum().sort_values(ascending=False).head(10).sort_values()
country_revenue.plot(kind="barh", ax=axes[1])
axes[1].set_title("Top 10 countries by revenue")
axes[1].set_xlabel("Revenue")

plt.tight_layout()
plt.show()
```

## RFM Feature Engineering

RFM is a common retail segmentation approach:

- **Recency:** days since the customer last purchased. Lower is better.
- **Frequency:** number of unique invoices. Higher is usually better.
- **Monetary:** total spending. Higher is usually better.

The reference date is set to one day after the latest transaction in the dataset, so Recency is measured consistently for all customers.

```python
# Build the RFM table: one row per customer instead of one row per transaction.

reference_date = df_clean["InvoiceDate"].max() + pd.Timedelta(days=1)

rfm = df_clean.groupby("CustomerID").agg(
    Recency=("InvoiceDate", lambda x: (reference_date - x.max()).days),
    Frequency=("InvoiceNo", "nunique"),
    Monetary=("TotalAmount", "sum"),
    TotalQuantity=("Quantity", "sum"),
    ProductDiversity=("StockCode", "nunique"),
).reset_index()

rfm["AvgOrderValue"] = rfm["Monetary"] / rfm["Frequency"]

print("Customer-level RFM shape:", rfm.shape)
rfm.head()
```

```python
# Review the scale and distribution of the customer-level RFM variables.

rfm[["Recency", "Frequency", "Monetary", "AvgOrderValue"]].describe().T
```

```python
# Plot RFM distributions to see skewness and possible outliers.

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].hist(rfm["Recency"], bins=40)
axes[0].set_title("Recency distribution")
axes[0].set_xlabel("Days since last purchase")

axes[1].hist(rfm["Frequency"], bins=40)
axes[1].set_title("Frequency distribution")
axes[1].set_xlabel("Number of invoices")

axes[2].hist(rfm["Monetary"], bins=40)
axes[2].set_title("Monetary distribution")
axes[2].set_xlabel("Total spend")

plt.tight_layout()
plt.show()
```

# STAGE 3 — Model & Evaluate

We model customer segments using K-Means clustering with RFM features.

Because K-Means is distance-based, the features must be scaled. The raw RFM variables are also highly skewed, especially `Frequency` and `Monetary`, so we apply `log1p` before scaling. This reduces the impact of extreme spenders while keeping the ranking and meaning of the variables.

Validation approach:

1. Split the customer table into train and test sets.
2. Fit the scaler and K-Means only on the train set.
3. Predict cluster labels for the test set.
4. Compare train and test silhouette scores.
5. Choose a value of `k` using both quantitative evidence and business interpretability.

```python
# Prepare the data for K-Means: split, transform skewed values and scale features.

FEATURES = ["Recency", "Frequency", "Monetary"]

rfm_train, rfm_test = train_test_split(
    rfm,
    test_size=0.20,
    random_state=RANDOM_SEED,
)

# log1p handles skew and works even when values are 0; here all values are positive.
X_train_log = np.log1p(rfm_train[FEATURES])
X_test_log = np.log1p(rfm_test[FEATURES])

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train_log)
X_test = scaler.transform(X_test_log)

print("Training customers:", len(rfm_train))
print("Test customers:", len(rfm_test))
```

```python
# Test several k values so the final number of clusters is not chosen blindly.

k_values = range(2, 9)
results = []

for k in k_values:
    model = KMeans(n_clusters=k, random_state=RANDOM_SEED, n_init=10)
    train_labels = model.fit_predict(X_train)
    test_labels = model.predict(X_test)

    results.append({
        "k": k,
        "train_inertia": model.inertia_,
        "train_silhouette": silhouette_score(X_train, train_labels),
        "test_silhouette": silhouette_score(X_test, test_labels),
        "train_test_gap": silhouette_score(X_train, train_labels) - silhouette_score(X_test, test_labels),
    })

k_results = pd.DataFrame(results)
k_results.round(4)
```

```python
# Visualize the elbow and silhouette results to support the choice of k.

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].plot(k_results["k"], k_results["train_inertia"], marker="o")
axes[0].set_title("Elbow method on train set")
axes[0].set_xlabel("Number of clusters (k)")
axes[0].set_ylabel("Inertia")

axes[1].plot(k_results["k"], k_results["train_silhouette"], marker="o", label="Train")
axes[1].plot(k_results["k"], k_results["test_silhouette"], marker="o", label="Test")
axes[1].set_title("Silhouette score by k")
axes[1].set_xlabel("Number of clusters (k)")
axes[1].set_ylabel("Silhouette score")
axes[1].legend()

plt.tight_layout()
plt.show()
```

## Choice of k

The evaluation table and the plots above compare several possible values of `k`. For this project, we choose `k = 3` because it matches the business question and gives three broad segments that are easy to explain to a marketing stakeholder:

1. **High-Value / Champions**
2. **At-Risk Customers**
3. **Low-Engagement Customers**

A larger value such as `k = 5` could create more detailed groups, but it would also make the recommendation harder to communicate in a 15-minute stakeholder presentation. Since this is a BI project for decision-making, we prioritize a simple and actionable segmentation. We still report the silhouette score honestly as the cluster-quality check.

The **At-Risk Customers** group is not an observed churn label. It is an interpretation based on weaker engagement than champions. The **Low-Engagement Customers** group is the segment with the weakest purchase activity overall.


```python
# Train the final K-Means model with k = 3, aligned with the business question.

K_FINAL = 3

kmeans = KMeans(n_clusters=K_FINAL, random_state=RANDOM_SEED, n_init=10)
kmeans.fit(X_train)

train_labels = kmeans.labels_
test_labels = kmeans.predict(X_test)

sil_train = silhouette_score(X_train, train_labels)
sil_test = silhouette_score(X_test, test_labels)

print(f"Final k: {K_FINAL}")
print(f"Train silhouette: {sil_train:.4f}")
print(f"Test silhouette : {sil_test:.4f}")
print(f"Train-test gap  : {sil_train - sil_test:+.4f}")
```

```python
# Assign each customer to one of the three final segments.

# Apply the fitted pipeline to every customer.
# The scaler and K-Means model are not re-fitted here.
X_all = scaler.transform(np.log1p(rfm[FEATURES]))
rfm["Cluster"] = kmeans.predict(X_all)

rfm["Cluster"].value_counts().sort_index()
```

```python
# Profile each cluster to understand how customers behave in each group.

cluster_profile = rfm.groupby("Cluster").agg(
    Customers=("CustomerID", "count"),
    Recency_mean=("Recency", "mean"),
    Frequency_mean=("Frequency", "mean"),
    Monetary_mean=("Monetary", "mean"),
    AvgOrderValue_mean=("AvgOrderValue", "mean"),
    ProductDiversity_mean=("ProductDiversity", "mean"),
).round(1)

cluster_profile
```

## Segment naming

K-Means returns numeric cluster labels such as 0, 1 and 2. These numbers do not have business meaning by themselves. The next cell converts the numeric clusters into three marketing names that match the business question:

- **High-Value / Champions:** best combination of recent purchases, frequency and monetary value.
- **At-Risk Customers:** not the strongest group, but still valuable enough to target with retention/reactivation actions.
- **Low-Engagement Customers:** weakest purchase activity, usually high recency and low frequency/monetary value.


```python
# Convert numeric cluster labels into business-friendly segment names.

cluster_means = rfm.groupby("Cluster")[FEATURES].mean()
cluster_z = cluster_means.apply(lambda s: (s - s.mean()) / s.std(ddof=0))

segment_names = {}

# Best segment: recent, frequent, high monetary value.
champion_cluster = (-cluster_z["Recency"] + cluster_z["Frequency"] + cluster_z["Monetary"]).idxmax()
segment_names[champion_cluster] = "High-Value / Champions"

# Weakest segment: old purchases, low frequency and low monetary value.
low_engagement_cluster = (cluster_z["Recency"] - cluster_z["Frequency"] - cluster_z["Monetary"]).idxmax()
segment_names[low_engagement_cluster] = "Low-Engagement Customers"

# The remaining cluster is interpreted as at-risk: not lost, but weaker than champions.
remaining = [c for c in cluster_means.index if c not in segment_names]
if remaining:
    segment_names[remaining[0]] = "At-Risk Customers"

rfm["Segment"] = rfm["Cluster"].map(segment_names)

pd.DataFrame({
    "Cluster": list(segment_names.keys()),
    "Segment": list(segment_names.values())
}).sort_values("Cluster")
```

```python
# Create a final segment summary table for interpretation and dashboarding.

segment_order = [
    "High-Value / Champions",
    "At-Risk Customers",
    "Low-Engagement Customers",
]

segment_summary = rfm.groupby("Segment").agg(
    Customers=("CustomerID", "count"),
    AvgRecency=("Recency", "mean"),
    AvgFrequency=("Frequency", "mean"),
    AvgMonetary=("Monetary", "mean"),
    TotalRevenue=("Monetary", "sum"),
    AvgOrderValue=("AvgOrderValue", "mean"),
    AvgProductDiversity=("ProductDiversity", "mean"),
).reindex(segment_order).round(1)

segment_summary["CustomerShare_%"] = (segment_summary["Customers"] / segment_summary["Customers"].sum() * 100).round(1)
segment_summary["RevenueShare_%"] = (segment_summary["TotalRevenue"] / segment_summary["TotalRevenue"].sum() * 100).round(1)

segment_summary
```

## Baseline comparison

Because this is an unsupervised segmentation problem, there is no true label and therefore no accuracy score. To avoid evaluating K-Means alone, we compare it with a simple **rule-based RFM baseline**.

The baseline creates customer groups using manual RFM scores. This represents a simple approach a marketing team could apply without machine learning. We then compare the baseline with K-Means using silhouette score and business interpretability.


```python
# Create a simple rule-based RFM baseline to compare against K-Means.

# Baseline segmentation using simple RFM rules.
# This is not a machine learning model; it is a manual business-rule comparison.

rfm["R_score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    q=3,
    labels=[3, 2, 1]
).astype(int)

rfm["F_score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    q=3,
    labels=[1, 2, 3]
).astype(int)

rfm["M_score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    q=3,
    labels=[1, 2, 3]
).astype(int)

rfm["Baseline_RFM_Score"] = rfm["R_score"] + rfm["F_score"] + rfm["M_score"]

rfm["Baseline_Segment"] = np.select(
    [
        rfm["Baseline_RFM_Score"] >= 8,
        rfm["Baseline_RFM_Score"] <= 4
    ],
    [
        "High-Value / Champions",
        "Low-Engagement Customers"
    ],
    default="At-Risk Customers"
)

rfm[["CustomerID", "Recency", "Frequency", "Monetary", "Baseline_RFM_Score", "Baseline_Segment"]].head()
```

```python
# Compare the model and the baseline using silhouette score.

# Compare K-Means against the rule-based RFM baseline.
# Both are evaluated on the SAME held-out test customers used in STAGE 3,
# so the comparison is fair and consistent with the silhouette already
# reported above (sil_test).

baseline_labels_test = pd.Categorical(
    rfm.loc[rfm_test.index, "Baseline_Segment"],
    categories=segment_order
).codes

baseline_silhouette_test = silhouette_score(X_test, baseline_labels_test)

baseline_comparison = pd.DataFrame({
    "Method": ["Rule-based RFM baseline", "K-Means clustering"],
    "Number of segments": [
        rfm["Baseline_Segment"].nunique(),
        rfm["Segment"].nunique()
    ],
    "Silhouette score": [
        baseline_silhouette_test,
        sil_test
    ],
    "Explanation": [
        "Simple manual segmentation using RFM scores",
        "Data-driven segmentation using scaled RFM features"
    ]
})

baseline_comparison.round(4)
```

```python
# Plot the baseline comparison in a simple chart for the report or slides.

baseline_comparison.set_index("Method")["Silhouette score"].plot(kind="bar", figsize=(7, 4))

plt.title("K-Means vs Rule-Based RFM Baseline")
plt.ylabel("Silhouette score")
plt.xlabel("")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

**Baseline interpretation:** The rule-based RFM baseline is useful because it is simple and easy to explain. K-Means is useful because it groups customers using the combined distance between Recency, Frequency and Monetary value instead of fixed manual cutoffs.

If K-Means has a similar or higher silhouette score than the baseline, it supports using the model. If the difference is small, we still keep K-Means because the final segments are coherent and directly connected to the marketing decision.

## Business Recommendation

**Main recommendation:** Use customer segments to personalize marketing instead of sending the same message to every customer.

- **High-Value / Champions:** Protect with loyalty benefits, early access and personalized service. Avoid unnecessary discounts because these customers already buy often and spend more.
- **At-Risk Customers:** Use retention and reactivation campaigns, such as personalized reminders, product recommendations and limited-time offers. This group is still worth targeting because it is not the weakest segment.
- **Low-Engagement Customers:** Use low-cost automated campaigns only, such as email reminders or simple promotions. Avoid spending large paid-marketing budgets on this group unless a small test proves it works.

# STAGE 5 — Ethics & Limits

**Privacy:** The dataset does not contain customer names, but `CustomerID` tracks purchasing behavior. If connected to a real CRM system, it would become personal or pseudo-anonymized data. Marketing teams should use it only with proper permissions and opt-in communication rules.

**Bias and representativeness:** The dataset is strongly concentrated in the United Kingdom, so the segments may not transfer directly to other countries or markets without retraining.

**Limitations:**

- The data covers a specific historical period, so customer behavior may have changed.
- K-Means assumes distance-based, roughly compact clusters. Real customers may not follow those mathematical shapes.
- The segmentation is descriptive, not causal. It does not prove that a campaign will increase revenue.
- The **At-Risk Customers** label is inferred from weaker purchase behavior than champions, not from an observed churn target.
- The **Low-Engagement Customers** label is based on low activity and low value; this does not mean those customers have no potential value.
- Removing returns/cancellations improves Monetary calculations, but it also removes information about product dissatisfaction or refund behavior.
- Choosing `k = 3` makes the segmentation easy to communicate, but it also hides some smaller subgroups that could appear with more clusters.
